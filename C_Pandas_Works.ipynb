# Pandas: Working with Works

This notebook is for use with the Melbourne Digital Musicology Workshops. It uses the Melbourne Rare Concert Programs dataset and focuses on one record type: **Works**.

Because the dataset contains several record types, not every field has a value in every row. Selecting Works first gives us a smaller, clearer table for learning how to inspect, filter, count, and sort data in Pandas.

## 1. Import Libraries

`requests` fetches the dataset from a URL, and `pandas` helps us work with it as a DataFrame.

```python
import requests
import pandas as pd
```

## 2. Load the Melbourne Concert Programs Dataset

Each item in the JSON describes one extracted fact from a concert program. The `record_type` field tells us what kind of fact it is. A Work record may have `composer`, `title`, and `movement_or_selection`, while other record types use different fields.

`pd.json_normalize()` turns the list of dictionaries into a flat DataFrame. When a field does not apply to a record, Pandas represents the missing value as `NaN`.

```python
url = 'https://raw.githubusercontent.com/RichardFreedman/Summer2026/e9e8a4a716f11d12a36a9c7db63c87503b1444aa/Melbourne_Rare_Concert/Structured%20Data/concert_program_items.json'
response = requests.get(url)
response.raise_for_status()
raw_data = response.json()
concerts = pd.json_normalize(raw_data)

concerts.head()
```

## 3. Select Only Work Records

A boolean condition creates a mask of `True` and `False` values. We use it to keep only rows where `record_type` is `work`.

This is useful here because fields such as `composer` and `title` are meaningful for Works but are often empty for other record types.

```python
works = concerts[concerts['record_type'] == 'work'].copy()
works.head(10)
```

### Keep the Most Useful Columns

A subset DataFrame contains only the columns we want to study.

```python
work_columns = [
    'record_type',
    'manifest_organization',
    'manifest_date',
    'page_number',
    'composer',
    'title',
    'movement_or_selection'
]
works = works[work_columns]
works.head(10)
```

### Reorder Columns

The order of columns in a DataFrame is controlled by the order of names in the list. Here we put the composer and title near the front because they are the main fields we want to study.

```python
reordered_columns = [
    'composer',
    'title',
    'movement_or_selection',
    'manifest_organization',
    'manifest_date',
    'page_number',
    'record_type'
 ]

works = works[reordered_columns]
works.head(10)
```

### Count Missing Values

`isna()` identifies missing values. Adding `.sum()` counts them in each column. This is especially important when working with records that do not all have the same fields.

Finding missing values is a **data-cleaning step**. After identifying them, we might decide to leave them missing, fill them with a meaningful value, remove rows, or investigate the original source. The right choice depends on the field and the research question. We will not make those cleaning decisions in this notebook; they belong in a later step, after the initial analysis.

```python
works.isna().sum()
```

## 5. Filter Works by Their Values

Use `.notna()` when you want only rows with a value in a particular field.

```python
named_composer_works = works[works['composer'].notna()]
named_composer_works[['composer', 'title']]
```

We can also look for a word in a title. `na=False` prevents missing titles from causing an error.

```python
works_with_symphony = works[
    works['title'].str.contains('symphony', case=False, na=False)
]
works_with_symphony[['composer', 'title']].head(10)
```

## 6. Count and Sort Values

`value_counts()` counts how often each value appears. By default, missing values are left out.

```python
works['composer'].value_counts().head(15)
```

```python
works['manifest_organization'].value_counts()
```

```python
works['manifest_date'].value_counts().sort_index()
```

To sort the most common composers alphabetically instead of by frequency, turn the counts into a Series and use `.sort_index()`.

```python
composer_counts = works['composer'].value_counts()
composer_counts.sort_index().head(15)
```

## 7. A Simple Data-Cleaning Example

The counts above are useful even when the data is not perfect. A separate cleaning step can create new, more convenient values for analysis. Here we create a surname-first composer column. We will split each name at the first space, reverse the two parts, and join them with a comma. This simple example assumes the composer value has a first name followed by a last name.

```python
# let's look at the _unique_ values for composers in order to understand the scope of names used
sorted(works['composer'].dropna().unique(), key=str.casefold)
```

### Some Basic Cleaning Steps

Clearly we can notice many different problems:

- Inconsistent values:  `L. v. Beethoven`, `L. V. Beethoven`, `L. van Beethoven`, `Beethoven`, `BEETHOVEN`, and so on
- Different naming conventions:  `Wagner, Richard`, `Richard Wagner`
- Missing values
- Wrong data type is possible (such as a `string` or `integer` in the place of a `date`)

And others, too! This is a typical problem, and each situation will require a different solution.  Indeed, some collections will require several steps to get your data to be consistent and clean!

Here we create a simple function that will `split` and `reverse` the name of a composer, so that `Richard Wagner` becomes `Wagner, Richard`.

- The function accepts any `string` we pass to it and returns the split and reversed items.  

- We `apply` function to all rows of the dataframe at once, and put the results in a new column.

This will NOT solve all the problems in our composer column, but it is a start!

```python
# create the function
def surname_first(composer):
    # a default value 'composer' if the field is empty
    if pd.isna(composer):
        return 'no composer given'
    # otherwise we split the name on the `space` between the names
    name_parts = composer.split(' ', 1)
    # and check the length of the result--if it's 2, then we return the two names separately, in reverse order
    if len(name_parts) == 2:
        return f'{name_parts[1]}, {name_parts[0]}'
    return composer


# finally we `apply` the function to the composer column, and return two new columns with the split names
works['composer_surname_first'] = works['composer'].apply(surname_first)
works[['composer', 'composer_surname_first']].drop_duplicates().head(50)
```
